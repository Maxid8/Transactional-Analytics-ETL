Get data from csv-s and manually check to readings were complete.  
Originally, I thought of reading the two csvs into a variable and then combining them with the concat() function. The problem with this is that the concat() function creates a separate copy when I join the two dataframes together. To avoid this, I read them in as a list, so the data only appears in memory once.  
The ignore_index is an important switch here, as joining the two csvs will prevent it from starting the indexing from the beginning.  
Since I looked at the raw csv files, I saw that they also contain dates. These must be explicitly provided to Pandas.

```python
import pandas as pd

df = pd.concat([pd.read_csv(f, parse_dates=['trans_date_trans_time', 'dob']) for f in ['fraudTest.csv', 'fraudTrain.csv']], ignore_index=True)
df.head()
```

Let's check for any duplicate rows.
Since Pandas looks at all the data in the row when checking for duplication and now we includes a column in our data that will not be needed (unnamed), it is worth filtering the duplications based on a subset, in this case based on the transaction numbers, as this must be unique.

```python
df = df.drop_duplicates(subset=['trans_num'])
df.shape
```
